```python
import requests 
import io
import zipfile 
import polars as pl
```

```python
response = requests.get(url="https://gtfs.edmonton.ca/TMGTFSRealTimeWebService/GTFS/gtfs.zip")

with zipfile.ZipFile(io.BytesIO(response.content)) as z: 
    stop_times_data = pl.read_csv(z.open("stop_times.txt"), infer_schema_length=None)
    routes_data = pl.read_csv(z.open("routes.txt"), infer_schema_length=None)
    trips_data = pl.read_csv(z.open("trips.txt"), infer_schema_length=None)
```
